```python
# General outlook
max_prob = np.max(probs)
min_prob = np.min(probs)
avg_prob = np.mean(probs)
median_prob = np.median(probs)

print("\n=== General outlook ===")
print(f"Maximum probability: {max_prob:.4f}")
print(f"Minimum probability: {min_prob:.4f}")
print(f"Average probability: {avg_prob:.4f}")
print(f"Median probability: {median_prob:.4f}")
```

```python
# Helper function to find contiguous regions above a threshold (length > 3 residues)
def find_regions_above_threshold(values, residues, positions, threshold):
    above = values > threshold
    regions = []
    i = 0
    n = len(values)
    while i < n:
        if above[i]:
            start = i
            while i < n and above[i]:
                i += 1
            end = i - 1
            length = end - start + 1
            if length > 3:   # more than 3 residues → at least 4
                region_residues = residues[start:end+1].tolist()
                region_positions = Position[start:end+1].tolist()
                region_vals = values[start:end+1]
                
                # Create a list of "position:residue" for easy printing
                pos_res_list = [f"{pos}:{res}" for pos, res in zip(region_positions, region_residues)]
                
                regions.append({
                    'start_position': Position[start],
                    'end_position': Position[end],
                    'length': length,
                    'residues': region_residues,           # list of residues
                    'positions': region_positions,         # list of positions
                    'pos_res_pairs': pos_res_list,         # formatted list
                    'probabilities': region_vals,          # numpy array or list
                    'cumulative_value': np.sum(region_vals),
                    'average_value': np.mean(region_vals)
                })
        else:
            i += 1
    return regions
```

```python
# Function to merge area dataframes with LepR_regions
def add_domain_info(area_df, lepr_df, keep_only_named_domains=True):
    """
    For each region in area_df (start_position, end_position), find all residues
    in lepr_df that fall inside that interval. Then attach:
      - domain(s) (region column from lepr_df)
      - region_position(s) (from lepr_df)
    Returns a new DataFrame with one row per region per unique domain.
    """
    merged_rows = []
    for idx, area_row in area_df.iterrows():
        r_start = area_row['start_position']
        r_end = area_row['end_position']
        
        # Find all LepR_regions rows with full_position in [r_start, r_end]
        mask = (lepr_df['full_position'] >= r_start) & (lepr_df['full_position'] <= r_end)
        overlapping = lepr_df[mask]
        
        if overlapping.empty:
            continue
        
        # If we only care about domains other than 'LeptinReceptor'
        if keep_only_named_domains:
            overlapping = overlapping[overlapping['region'] != 'LeptinReceptor']
            if overlapping.empty:
                continue
        
        # Group by domain (region) to collect region_positions and residues
        grouped = overlapping.groupby('region')
        for domain_name, group in grouped:
            region_positions = group['region_position'].tolist()
            # Optionally collect residues as well
            residues_in_domain = group['residue'].tolist()
            
            new_row = area_row.to_dict()
            new_row['domain'] = domain_name
            new_row['region_positions'] = ';'.join(map(str, region_positions))
            new_row['domain_residues'] = ''.join(residues_in_domain)
            new_row['num_residues_in_domain'] = len(group)
            merged_rows.append(new_row)
    
    result = pd.DataFrame(merged_rows)
    if not result.empty:
        # Reorder columns to put domain info near front
        cols = ['domain', 'region_positions', 'domain_residues', 'num_residues_in_domain'] + \
               [c for c in result.columns if c not in ['domain', 'region_positions', 'domain_residues', 'num_residues_in_domain']]
        result = result[cols]
    return result
```
