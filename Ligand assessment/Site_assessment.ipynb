```python
# import modules
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os
from google.colab import drive
# make sure its using colab to connect with google drive. dont know how to do it in other way, yet, anyway?
```

```python
# Assessment data data
Assessment_score = 'https://github.com/FairzanIsLearning/OralLeptinExploration/blob/main/Ligand%20assessment/OG_LepFr_LepR.csv?raw=true'
Fitting_rawdata = pd.read_csv(Assessment_score)

# print(Fitting_rawdata.head())
```

```python
# Extract data
residues = Fitting_rawdata["Residue"]
probs = Fitting_rawdata["Probabilities"]
Position = Fitting_rawdata["Position"]
```

```python
# General outlook
max_prob = np.max(probs)
min_prob = np.min(probs)
avg_prob = np.mean(probs)
median_prob = np.median(probs)

print("\n=== General outlook ===")
print(f"Maximum probability: {max_prob:.4f}")
print(f"Minimum probability: {min_prob:.4f}")
print(f"Average probability: {avg_prob:.4f}")
print(f"Median probability: {median_prob:.4f}")
```

```python
# Helper function to find contiguous regions above a threshold (length > 3 residues)
def find_regions_above_threshold(values, residues, positions, threshold):
    above = values > threshold
    regions = []
    i = 0
    n = len(values)
    while i < n:
        if above[i]:
            start = i
            while i < n and above[i]:
                i += 1
            end = i - 1
            length = end - start + 1
            if length > 3:   # more than 3 residues → at least 4
                region_residues = residues[start:end+1].tolist()
                region_positions = Position[start:end+1].tolist()
                region_vals = values[start:end+1]
                
                # Create a list of "position:residue" for easy printing
                pos_res_list = [f"{pos}:{res}" for pos, res in zip(region_positions, region_residues)]
                
                regions.append({
                    'start_position': Position[start],
                    'end_position': Position[end],
                    'length': length,
                    'residues': region_residues,           # list of residues
                    'positions': region_positions,         # list of positions
                    'pos_res_pairs': pos_res_list,         # formatted list
                    'probabilities': region_vals,          # numpy array or list
                    'cumulative_value': np.sum(region_vals),
                    'average_value': np.mean(region_vals)
                })
        else:
            i += 1
    return regions
```

```python
# Find regions above average
print("\n=== Regions with >3 residues above AVERAGE ===")
regions_avg = find_regions_above_threshold(probs, residues, Position, avg_prob)
Area_above_avg = pd.DataFrame.from_dict(regions_avg)

if not regions_avg:
    print("No such regions found.")
else:
    for idx, reg in enumerate(regions_avg, 1):
        print(f"\nRegion {idx}:")
        print(f"  Length: {reg['length']} residues")
        print(f"  Residues (with positions): {', '.join(reg['pos_res_pairs'])}")
        print(f"  Cumulative probability: {reg['cumulative_value']:.4f}")
        print(f"  Average probability: {reg['average_value']:.4f}")

Area_above_avg.to_csv(os.path.join(folder_path, 'regions_above_average.csv'), index=False)
```

```python
# Find regions above median
print("\n=== Regions with >3 residues above MEDIAN ===")
regions_med = find_regions_above_threshold(probs, residues, Position, median_prob)
Area_above_med = pd.DataFrame.from_dict(regions_med)

if not regions_med:
    print("No such regions found.")
else:
    for idx, reg in enumerate(regions_med, 1):
        print(f"\nRegion {idx}:")
        print(f"  Length: {reg['length']} residues")
        print(f"  Residues (with positions): {', '.join(reg['pos_res_pairs'])}")
        print(f"  Cumulative probability: {reg['cumulative_value']:.4f}")
        print(f"  Average probability: {reg['average_value']:.4f}")

Area_above_med.to_csv(os.path.join(folder_path, 'regions_above_median.csv'), index=False)
```
