```python
#sequence converter to SMILES or image

def BILN_generate(FASTA):
    separator = '-'
    res = ''
    for i in range(0, len(FASTA), 1):
        res += FASTA[i:i + 1] + separator
    # remove the trailing separator
    res = res[:-1]
    return res

def FASTA_to_SMILES(sequence):
    biln = BILN_generate(sequence)
    romol = Molecule(Sequence(biln)).get_molecule(fmt='ROMol')
    return Chem.MolToSmiles(romol)

def FASTA_to_2Dstructure(sequence):
    biln = BILN_generate(sequence)
    romol = Molecule(Sequence(biln)).get_molecule(fmt='ROMol')
    return romol
```
