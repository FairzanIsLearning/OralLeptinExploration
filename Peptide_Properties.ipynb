```python
# Peptides.py module https://peptides.readthedocs.io/en/stable/api/peptide.html
from peptides import Peptide
```

```python
# Inputted the peptide sequences
LepMotif = Peptide("CHLPW")
#-------------
LepMain = Peptide("SCHLPWASGLETLDS")
Seq1 = Peptide("SCHLPW")
Seq2 = Peptide("CHLPWA")
Seq2mod1 = Peptide("CHLPWV")
Seq2mod2 = Peptide("CHLPWS")
Seq2mod3 = Peptide("CHLPWM")
peptides_list = [LepMain, Seq1, Seq2, Seq2mod1, Seq2mod2, Seq2mod3]
```

```python
# Calculate and print hydrophobicity for each
for i, pep in enumerate(peptides_list):
    hydrophobicity_value = pep.hydrophobicity(scale='Eisenberg')
    print(f"Peptide {i+1}: {pep.sequence} -> hydrophobicity = {hydrophobicity_value:.3f}")
```

```python
# Calculate and print instability for each
for i, pep in enumerate(peptides_list):
    instability_value = round(pep.instability_index(), 2)
    print(f"Peptide {i+1}: {pep.sequence} -> instability index = {instability_value:.2f}")
```
